## Simple Arithmetic Calculator

**Building a simple command-line calculator that performs basic arithmetic operations such as addition, subtraction, mulitplication, and division.**

```python
def add(x, y):
    return x + y
def subtract(x, y):
    return x - y
def multiply(x, y):
    return x * y
def divide(x, y):
    if y == 0:
        return "Error: Enter a non-zero" #Error handling
    return x / y

user_name = input("What is your name: ") # collects the user's name
print("")

s1 = "**************************************************************"
s2 = "*                                                            *"
print(f"{s1}\nWelcome {user_name}, this is a simple arithmetic calculator \n{s1}")
print("")
print("Choose operation:")
print("1. Add(+)")
print("2. subtract(-)")
print("3. Multiply(*)")
print("4. Divide(/)")

optn = input("Enter your desired operator: ")
a = float(input("Enter the first number: "))
b = float(input("Enter the second number: "))

if optn == '+':
    result = add(a, b) # adds a and b
elif optn == '-':
    result = subtract(a,b) # subtract b from a 
elif optn == '*':
    result = multiply(a, b) # multiply a with b
elif optn == '/':
    result = divide(a, b) # remove b from a
else:
    result = "Invalid operator"

print('Result: ', result)
```
